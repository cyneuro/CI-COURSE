## 🧭 CI-HARDWARE Group 1: Core Architecture Fundamentals

### Concept 1 — CPU (Central Processing Unit)

**Resource:** The chip that executes instructions — fetch, decode, execute, repeat, billions of times per second. A modern consumer CPU runs at roughly 3–5 GHz (billions of cycles per second), with most laptop/desktop CPUs having 4–16 cores as of 2026.



### Concept 2 — Core

**Resource:** An independent processing unit within a single CPU chip. A "quad-core" CPU can genuinely execute 4 separate instruction streams at once — this is the actual hardware basis for the parallel speedup you computed back in Colab 2.



### Concept 3 — RAM (Random Access Memory)

**Resource:** Fast, temporary working memory — everything a running program is actively using lives here, and it's wiped when power is lost. A typical laptop has 16–32 GB in 2026; a micro:bit, by contrast, has 128 KB — roughly 250,000 times less.



### Concept 4 — Cache

**Resource:** A small pool of extremely fast memory sitting between the CPU and RAM, storing recently-used data so the CPU doesn't have to wait for slower RAM every time. Access to cache can be 10–100x faster than access to RAM.



### Concept 5 — Clock Speed (GHz)

**Resource:** How many cycles per second a CPU can execute — 1 GHz = 1 billion cycles/second. Higher clock speed generally means faster execution *per core*, but it's only half the performance picture; core count is the other half.



### Concept 6 — 32-bit vs. 64-bit Architecture

**Resource:** The width of data a processor handles in one operation, and critically, how much RAM it can even address. A 32-bit system can address a maximum of about 4 GB of RAM — which is why virtually all modern systems (with 16+ GB RAM) are 64-bit.



### Concept 7 — Storage vs. Memory

**Resource:** RAM is fast and temporary; storage (an SSD or hard drive) is slower but permanent. A typical 2026 laptop SSD reads data at roughly 3,000–7,000 MB/s — fast, but still far slower than RAM's effectively instant access.



### Concept 8 — Motherboard and Bus

**Resource:** The bus is the physical pathway data travels between the CPU, RAM, and storage — its bandwidth is a real, often-overlooked bottleneck. A modern PCIe 5.0 bus moves data at roughly 63 GB/s per lane, far more than earlier generations, but still finite.



### Concept 9 — GPU vs. CPU Architecture

**Resource:** A CPU has a handful of powerful, flexible cores (4–16, typical); a GPU has thousands of small, simple cores (a 2026 consumer GPU might have 8,000–16,000). That's why GPUs excel at the same simple operation repeated many times — like the matrix math inside Colab 2's CNN — and CPUs excel at varied, sequential logic.



### Concept 10 — Moore's Law and Its Limits

**Resource:** The historical observation that transistor density roughly doubled every two years, driving decades of performance gains. As of the mid-2020s, this pace has genuinely slowed due to physical limits on transistor size — a real reason the industry has shifted toward adding cores and specialized chips (GPUs, neuromorphic hardware) rather than just faster single cores.

## 🧭 CI-HARDWARE Group 2: Edge/Embedded Devices

**1. Microcontroller vs. Microprocessor** — A microcontroller (like the micro:bit's) packages a CPU, RAM, and storage onto one chip; a microprocessor (like a laptop's CPU) is just the processing core, requiring separate RAM and storage chips. This is *why* the micro:bit fits so much into so little: 128 KB isn't a limitation of a stripped-down laptop chip, it's the actual design.



**2. Embedded System** — A computer built into a device to perform one specific job, not general-purpose computing. Your car, microwave, and the micro:bit are all embedded systems; your laptop isn't.



**3. Real-Time Operating System (RTOS) vs. General OS** — An RTOS (common on embedded devices) guarantees a task completes within a fixed time window; a general OS (Windows, macOS) prioritizes overall responsiveness, not hard timing guarantees. This distinction matters directly for closed-loop control — Track B's "Latency in Closed-Loop Control" concept from Colab 3 is exactly why some embedded systems need an RTOS.



**4. Memory-Mapped I/O** — On constrained devices, sensors and outputs are often addressed as if they were memory locations, letting simple read/write operations control physical hardware. This is part of what makes 128 KB of RAM go further than it sounds.



**5. Sensor/Actuator Interfacing (ADC/DAC)** — Real-world signals (temperature, light) are analog; a computer only understands digital values. An Analog-to-Digital Converter (ADC) is the actual hardware bridging that gap — every "sensor reading" you simulated in Colab 3 stands in for this real conversion step.



**6. Power Constraints in Edge Devices** — A micro:bit can run for weeks on a coin-cell battery; a laptop lasts hours. This isn't incidental — it directly limits what computation is even possible at the edge, which is part of why edge devices favor simple, low-power operations over anything computationally heavy.



**7. Firmware** — Software permanently (or semi-permanently) embedded into a device's hardware, sitting below the operating system. Updating a device's firmware is a fundamentally different, riskier operation than installing an app — a failed firmware update can brick the device entirely.



**8. Edge Computing vs. Cloud Computing** — Processing data locally, on the device collecting it (edge), vs. sending it elsewhere for processing (cloud). The real trade-off is latency and bandwidth: edge computing avoids network delay entirely, at the cost of the device's limited compute power.



**9. Single-Board Computers (e.g., Raspberry Pi) vs. Microcontrollers** — A Raspberry Pi runs a full operating system (Linux) and has far more RAM (1–8 GB) than a micro:bit's 128 KB — a genuinely different tier, sitting between "edge device" and "laptop."



**10. IoT (Internet of Things) Device Constraints** — Networked embedded devices face the same power/memory limits as any edge device, plus the added cost of maintaining a network connection — often the single most power-hungry component on the device.

## 🧭 CI-HARDWARE Group 3: Personal/Departmental Computing

**1. Laptop vs. Desktop Architecture** — Functionally similar, but laptops trade raw performance for power efficiency and portability — a desktop CPU of the same generation typically runs at a higher sustained clock speed since it isn't power- or heat-constrained the same way.



**2. The OS's Role in Resource Management** — The operating system decides which program gets CPU time, when, and how much RAM each gets — this is genuinely a scheduling problem, at a much smaller scale than Slurm's cluster-wide job scheduling in Colab 3.



**3. Virtual Memory / Swap Space** — When RAM fills up, the OS can temporarily move some data to disk, effectively extending memory at the cost of speed (disk is far slower than RAM). Heavy swap use is a real, common cause of a computer "freezing" under load.



**4. Multi-Tasking and Process Scheduling** — Even a "single-core" feel from decades ago was an illusion — the OS was rapidly switching between programs. Modern multi-core CPUs make this less of a trick and more literal, but the OS-level scheduling problem remains real.



**5. Departmental/Lab Server** — A shared machine (typically 32–128 cores, far more RAM than a laptop) serving one lab or department — the tier directly above "laptop" on the CI-Hardware ladder, often the first real step up for a growing research project.



**6. Network-Attached Storage (NAS)** — Shared storage accessible over a local network rather than living inside any one machine, letting a whole lab access the same dataset without copying it to every laptop.



**7. RAID (Redundant Array of Independent Disks)** — Multiple physical drives combined to improve speed, reliability, or both — a lab server holding years of research data typically uses RAID specifically so a single drive failure doesn't mean data loss.



**8. Local Area Network (LAN) Basics** — The network connecting machines within one building or department, typically far faster and lower-latency than the internet connection reaching outside it.



**9. Remote Desktop / SSH Basics** — Ways of controlling a machine you're not physically sitting at — SSH (command-line) is the same underlying mechanism behind VSCode's Remote-SSH from Colab 1, just used directly.



**10. Total Cost of Ownership: Personal vs. Shared** — A departmental server costs more upfront than a laptop, but split across an entire lab's usage, often costs less per researcher than everyone buying individually powerful personal machines.

## 🧭 CI-HARDWARE Group 4: Institutional/Shared Computing

**1. HPC Cluster Architecture (Head Node vs. Compute Nodes)** — Users log into a single head/login node, but their actual jobs run on separate compute nodes — you never do heavy computation directly on the node you log into, exactly the pattern FABRIC's own access model follows.



**2. Interconnect (e.g., InfiniBand)** — The specialized, low-latency networking connecting nodes within a cluster — dramatically faster than standard networking, which matters enormously for Colab 3's multi-node jobs, where nodes must exchange data mid-computation.



**3. Shared Filesystem Across a Cluster** — All compute nodes typically see the exact same file system, so a job can start on any node and access the same data without manual copying — this is what makes Colab 3's distributed circuit model practically workable.



**4. Node Allocation and Partitioning** — Clusters are often split into partitions (e.g., a GPU partition, a high-memory partition) so Slurm can match a job's request to hardware that actually has what it needs.



**5. Queueing and Fair-Share Scheduling** — Revisiting Slurm from the hardware-capacity side: fair-share scheduling deliberately lowers the priority of users who've already consumed a lot of recent cluster time, so no single user can permanently monopolize shared hardware.



**6. High-Memory Nodes vs. Standard Nodes** — Most cluster nodes carry standard RAM (64–256 GB); a smaller number of "high-memory" nodes might carry 1 TB or more, reserved for jobs that specifically need it.



**7. GPU Nodes Within a Cluster** — Not every node in a cluster has a GPU — they're a distinct, often scarcer resource pool, which is exactly why Colab 2's `#SBATCH` directives need to specifically request GPU access, not assume it.



**8. Cluster Utilization Metrics** — Institutions track what fraction of total cluster capacity is actually in use — a cluster sitting at 40% utilization represents real wasted institutional investment, part of why "request only what you need" (Shared Cluster Etiquette, Colab 3) genuinely matters at scale.



**9. Total FLOPS as a Cluster-Level Metric** — Floating-point operations per second, summed across every node, is the standard way institutions report a cluster's total computing capacity — a real, comparable number, the same unit ACCESS uses when describing its national-scale capacity.



**10. Cost Recovery Models** — Many universities charge researchers' grants for HPC allocation time rather than funding it as pure overhead — a real economic reason "request only what you need" isn't just etiquette, it's often literally billed.

## 🧭 CI-HARDWARE Group 5: National/Distributed Infrastructure

**1. FABRIC, Revisited** — Beyond "a national testbed": FABRIC is specifically a *programmable, software-defined* infrastructure — researchers can reconfigure the network itself as part of an experiment, not just run computation on fixed hardware.

ACCESS, Revisited** — Coordinates allocations across multiple national supercomputing centers at once, collectively operating at petaflop scale — a quadrillion floating-point operations per second, a number this course has cited before but worth actually sitting with: that's roughly a million laptops' worth of raw computation, combined.

**5. **9. Serverless Computing, Revisited** — At the infrastructure level: serverless doesn't mean "no servers," it means the *provider* manages server allocation dynamically, invisible to the user — you request compute, not a specific machine, which is the real mechanism behind Colab 3's scale-to-zero pricing.

Neuromorphic and Quantum Computing, Revisited** — Both represent a genuine architectural break from conventional (von Neumann) computing, not just faster versions of it — DOE and EU roadmaps place large-scale, fault-tolerant versions of both on a 10–20 year horizon, a real research trajectory, not speculation.

## 🧭 CI-SOFTWARE Group 1: Machine-Level Execution

### Concept 1 — Binary / Machine Code

**Resource:** The only language a processor genuinely executes — sequences of 0s and 1s directly interpreted by hardware circuitry. Every layer of software you've used all semester ultimately compiles or interprets down to this.



### Concept 2 — Assembly Language

**Resource:** A thin, human-readable layer directly above machine code — one assembly instruction typically maps to exactly one machine instruction. Writing in assembly means managing individual registers and memory addresses by hand, which is why almost no one writes application code in it today.



### Concept 3 — Compiler

**Resource:** A program that translates an entire high-level source file (like C or Fortran) into machine code *before* the program runs. Compiled languages typically run faster than interpreted ones because this translation work happens once, upfront, not repeatedly during execution.



### Concept 4 — Interpreter

**Resource:** A program that translates and executes source code line-by-line, at runtime, rather than all at once beforehand. Python is interpreted — which is part of why it's slower than compiled C for the same raw computation, but faster to write and iterate on, a real trade-off, not a flaw.



### Concept 5 — Source Code vs. Executable

**Resource:** Source code is the human-readable file you write; an executable is the machine-code file that actually runs. A compiled language produces a genuinely separate executable file; an interpreted language like Python typically has no separate executable at all — the interpreter reads the source directly.



### Concept 6 — Instruction Set Architecture (ISA)

**Resource:** The specific vocabulary of operations a given processor understands — x86 (most laptops/servers) and ARM (most phones, and Apple's recent Macs) are two major, mutually incompatible ISAs. This is a real, practical fact: an executable compiled for one ISA generally won't run on the other without recompilation or emulation.



### Concept 7 — Opcode

**Resource:** The specific numeric code within a machine instruction that tells the CPU which operation to perform (add, load, jump, etc.). Every single line of Python you've written this semester ultimately becomes thousands of these.



### Concept 8 — Linker

**Resource:** The tool that combines multiple separately-compiled pieces of code (your code, plus the libraries it uses) into one final executable. This is a genuinely separate step from compiling — a common real-world error ("undefined reference") happens at link time, not compile time.



### Concept 9 — Why We Don't Hand-Write Assembly Today

**Resource:** Not because it's impossible, but because of a real productivity trade-off: a task taking a few lines of Python can take dozens of lines of assembly, with far more room for error. This is the first rung of the exact abstraction ladder Colab 1 introduced, ending (for now) at vibe coding.



### Concept 10 — Machine Language's Real Limitation

**Resource:** Machine code is tied to one specific ISA — code written for x86 literally cannot run on ARM hardware without translation. Every higher layer of software (compilers, interpreters, virtual machines) exists partly to solve exactly this portability problem.

## 🧭 CI-SOFTWARE Group 2: Language Evolution

**1. High-Level Programming Language** — A language (Python, Java) that abstracts away hardware-level detail, letting a programmer express intent without managing memory addresses or registers directly — the entire point of every abstraction layer above assembly.



**2. Object-Oriented Programming (OOP)** — Organizing code around "objects" that bundle data and the functions that act on it together, rather than treating data and logic as entirely separate — Python's classes (used throughout Colab 3 and 4's device/robot code) are a direct application of this.



**3. Procedural vs. Object-Oriented vs. Functional Paradigms** — Three different ways of structuring a program's logic: procedural (a sequence of steps), object-oriented (bundled data+behavior), functional (composing pure functions with no shared mutable state). Most real Python code, including this course's, mixes procedural and object-oriented freely.



**4. Python's Design Philosophy** — Explicitly prioritizes readability and "batteries included" (a large standard library) over raw execution speed — a deliberate trade-off, not an oversight, and part of why it dominates research computing despite being slower than compiled alternatives for raw computation.



**5. Dynamic Typing vs. Static Typing** — Python decides a variable's type at runtime (dynamic); languages like C require declaring it upfront (static). Dynamic typing is faster to write, but a whole category of bugs (assigning the wrong type) that a static-typed compiler would catch immediately can instead surface only when that specific line finally runs.



**6. Garbage Collection / Automatic Memory Management** — Python automatically reclaims memory no longer in use; older languages like C require the programmer to explicitly free it, and forgetting to is a classic, serious bug category ("memory leaks") that automatic garbage collection genuinely eliminates.



**7. Standard Library vs. Third-Party Packages** — Python ships with a large built-in standard library (e.g., `os`, `datetime`) at no extra install; `pandas`, `torch`, and everything else you `pip install`ed this semester are third-party packages, maintained outside Python itself.



**8. Package Managers (pip, conda)** — Tools that install, update, and track dependencies for you — this is the actual mechanism behind Colab 2's `pip install` and `conda create` commands, not just convenient shorthand.



**9. Scripting Languages vs. Systems Languages** — Python (scripting) prioritizes fast iteration; C/Rust (systems languages) prioritize raw performance and low-level hardware control. Real production systems often use both together — a Python interface calling into performance-critical C code underneath, which is literally how libraries like NumPy and PyTorch are built.



**10. Domain-Specific Languages (DSLs)** — A language built for one narrow purpose rather than general programming — SQL (databases) and regular expressions (pattern matching) are both DSLs you've likely encountered without necessarily naming them as such.

## 🧭 CI-SOFTWARE Group 3: Environments & Dependency Management

**1. Dedicated Software Environments, Revisited** — Beyond isolation for its own sake: a dedicated environment is what makes a documented environment file *meaningful* — without isolation, "here's what I installed" doesn't reliably describe what your code actually runs against.



**2. Environment Files, Revisited** — The actual artifact of reproducibility: a well-formed `environment.yml` or `requirements.txt` is, in a real sense, a complete recipe for recreating your exact software conditions, not just a convenience for sharing.



**3. Package Version Conflicts ("Dependency Hell")** — When two packages you need require incompatible versions of a shared dependency, no configuration can satisfy both simultaneously in one environment — a real, common, sometimes genuinely unsolvable problem within a single environment, which is part of why separate environments per project exist at all.



**4. Virtual Environments (venv) vs. Conda Environments** — Python's built-in `venv` isolates Python packages only; `conda` environments can additionally manage non-Python dependencies (compilers, system libraries) — part of why Colab 2's FABRIC setup used conda rather than venv.



**5. Containers (Docker), Revisited** — Beyond "packages code and libraries together": a container also includes a minimal operating system layer, meaning it can guarantee identical behavior across genuinely different host machines in a way an environment file alone cannot.



**6. Container Registries** — Public or private repositories (like Docker Hub) hosting pre-built container images — a lab can publish its exact computational environment for anyone to pull and run, a real, increasingly common practice for reproducible research.



**7. Version Control (Git), Revisited** — Beyond "tracks changes": Git specifically tracks changes as a history of discrete, reversible commits, meaning a bad change can be identified and undone precisely, not just guessed at.



**8. Git Branches and Merge Conflicts** — A branch lets you develop a change in isolation before merging it back into the main codebase; a merge conflict happens when two branches changed the same lines differently, and Git genuinely cannot resolve it automatically — a person has to decide.



**9. Continuous Integration (CI/CD) Basics** — Automatically running tests every time code changes, catching a broken change within minutes rather than whenever someone happens to notice — a real, standard practice in professional software development this course hasn't directly used, but which underlies how reliable open-source tools (like the ones you `pip install`) stay reliable.



**10. Software Licensing** — Open-source licenses (MIT, GPL, and others) genuinely differ in what they permit — MIT allows nearly unrestricted reuse, GPL requires derivative works to also be open-sourced. Which license a tool uses is a real, practical constraint on how you're allowed to use it, not just fine print.

## 🧭 CI-SOFTWARE Group 4: Distributed/Parallel Software

**1. Parallel Programming Models: Shared-Memory vs. Distributed-Memory** — Shared-memory parallelism (multiple cores on one machine) lets threads directly access the same RAM; distributed-memory parallelism (multiple separate machines) requires explicit communication, since no memory is actually shared — a genuinely harder programming problem.



**2. Message Passing Interface (MPI)** — The standard protocol for distributed-memory communication between nodes — when Colab 3's multi-node circuit model "combines results" across nodes, real HPC code typically does this via MPI, even though our simulation abstracted that detail away.



**3. Slurm, Revisited** — Beyond "the job scheduler": Slurm's directives (`--nodes`, `--time`, `--mem`) are literally a contract — you're telling the scheduler exactly what your job needs, and it won't run until hardware matching that contract becomes available.



**4. Batch Processing vs. Interactive Computing** — Batch (submit a job, walk away, results ready later) trades immediacy for the ability to use resources efficiently and unattended; interactive computing (like a live Colab session) trades resource efficiency for immediate feedback — both are legitimate, and this course has used both deliberately at different points.



**5. Race Conditions and Synchronization** — When multiple parallel processes access shared data without coordination, the final result can depend on unpredictable timing — a genuinely nondeterministic bug class, and part of why Colab 3's job-dependency logic (waiting for all nodes before combining) exists at all.



**6. Load Balancing Across Nodes** — Distributing work unevenly (giving more to faster nodes, less to slower ones) versus evenly — real clusters often have heterogeneous hardware, making load balancing a genuine optimization problem, not just an even split.



**7. MapReduce / Distributed Data Processing** — A pattern for processing huge datasets: "map" splits work across many machines, "reduce" combines the results — conceptually, exactly the shape of Colab 3's multi-node circuit model, generalized to a much larger, industry-standard pattern.



**8. Fault Tolerance in Distributed Systems** — At large scale, some node failing during a long job isn't a rare edge case, it's close to guaranteed — real distributed systems are designed assuming failure will happen, not hoping it won't.



**9. Checkpointing** — Periodically saving a long-running job's progress so a failure doesn't mean starting over from zero — directly relevant to Colab 3's wall-clock time limit: without checkpointing, a job killed at its time limit loses everything since the last save.



**10. Scalability: Horizontal vs. Vertical** — Vertical scaling means a bigger single machine; horizontal scaling means more machines working together. This course's own hardware ladder (laptop → server → cluster → national resource) is fundamentally a horizontal-scaling story, not a vertical one.
